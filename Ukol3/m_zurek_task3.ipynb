```python
import numpy as np
import matplotlib.pyplot as plt

P = 3
Fsig = 1
Amp = 1
DC = 2
Fs = 1000

Ts = 1 / Fs
N = int((P / Fsig) * Fs)
k = np.arange(N)
t = k * Ts

y_sig = DC + Amp * np.sin(2 * np.pi * Fsig * t)

U_RMS_sig = np.sqrt(DC**2 + (Amp**2) / 2)

U_RMS_noise_10 = U_RMS_sig / (10**(10 / 20))
U_RMS_noise_30 = U_RMS_sig / (10**(30 / 20))

noise_10 = np.random.randn(N) * U_RMS_noise_10
noise_30 = np.random.randn(N) * U_RMS_noise_30

y_super_10 = y_sig + noise_10
y_super_30 = y_sig + noise_30
```

```python
fig1, ax1 = plt.subplots(figsize=(10, 5))
ax1.plot(t, y_sig, 'k-', label='Signal')
ax1.plot(t, noise_10, 'r-', label='Noise', alpha=0.7)

title_1 = r'Signal and noise separately shown (desired SNR: 10 dB)' + '\n' + r'$F_{sig}=1$ Hz, Amp=1 V, DC=2 V, $F_s=1$ kHz'
ax1.set_title(title_1)
ax1.set_xlabel('Time [s]')
ax1.set_ylabel('Voltage [V]')

ax1.set_xlim(0, 3)
ax1.set_ylim(-3, 3)

xticks1 = np.arange(0, 3.1, 0.5)
ax1.set_xticks(xticks1)
ax1.set_xticklabels([f"{x:g}" for x in xticks1])

yticks1 = np.arange(-3, 3.1, 1)
ax1.set_yticks(yticks1)
ax1.set_yticklabels([f"{y:g}" for y in yticks1])

ax1.tick_params(direction='in', top=True, right=True, bottom=True, left=True)
ax1.legend(loc='upper right')
plt.show()

fig2, ax2 = plt.subplots(figsize=(10, 5))
ax2.hist(noise_10, bins=50, color='b', edgecolor='black', alpha=0.7)

title_2 = r'Frequency distribution histogram'
ax2.set_title(title_2)
ax2.set_xlabel('Voltage [V]')
ax2.set_ylabel('Frequency [-]')

ax2.tick_params(direction='in', top=True, right=True, bottom=True, left=True)
plt.show()

xticks2 = np.arange(-3, 3.1, 0.5)
ax2.set_xticks(xticks2)
ax2.set_xticklabels([f"{x:g}" for x in xticks2])

ax2.tick_params(direction='in', top=True, right=True, bottom=True, left=True)
plt.show()

U_RMS_sig_act = np.sqrt(np.mean(y_sig**2))
U_RMS_noise_10_act = np.sqrt(np.mean(noise_10**2))
SNR_10_act = 20 * np.log10(U_RMS_sig_act / U_RMS_noise_10_act)

fig3, ax3 = plt.subplots(figsize=(10, 5))
ax3.plot(t, y_super_10, 'b-', label=f'Superimposed noise (Real SNR: {SNR_10_act:.1f} dB)')

title_3 = r'Superimposed noise (Real SNR: 10 dB)'
ax3.set_title(title_3)
ax3.set_xlabel('Time [s]')
ax3.set_ylabel('Voltage [V]')

ax3.set_xlim(0, 3)
ax3.set_ylim(-2, 6)

xticks3 = np.arange(0, 3.1, 0.5)
ax3.set_xticks(xticks3)
ax3.set_xticklabels([f"{x:g}" for x in xticks3])

yticks3 = np.arange(-2, 6.1, 1)
ax3.set_yticks(yticks3)
ax3.set_yticklabels([f"{y:g}" for y in yticks3])

ax3.tick_params(direction='in', top=True, right=True, bottom=True, left=True)
ax3.legend(loc='upper right')
plt.show()
```

## Komentář

Nejprve jsem importoval nezbytné knihovny. Následně jsem definoval zadané proměnné: počet period P = 3, frekvenci signálu f<sub>sig</sub> = 1 Hz, vzorkovací frekvenci f<sub>s</sub> = 1 kHz, amplitudu Amp = 1 V a stejnosměrnou složku DC = 2 V. Z těchto hodnot jsem odvodil časový krok T<sub>s</sub> a celkový počet vzorků $N$ pro vytvoření časové osy.

Sinusový signál jsem vygeneroval dle zadaných parametrů. Aby bylo možné přimíchat správné množství šumu, nejprve jsem analyticky vypočítal efektivní hodnotu signálu (U<sub>RMSsignal</sub>) ze vzorce využívajícího stejnosměrnou složku a amplitudu. Následně jsem z definice odstupu signál-šum odvodil efektivní hodnotu šumu (U<sub>RMSnoise</sub>) pro případ 10 dB a případ 30 dB. Samotný šumový signál jsem vygeneroval pomocí funkce `numpy.random.randn`, jež zajišťuje ideální normální Gaussovo rozložení s nulovou střední hodnotou. Tento základní šum jsem poté škáloval přenásobením vypočtenou směrodatnou odchylkou.

Pro každý případ (10 dB i 30 dB) jsem vygeneroval tři grafy. První graf vždy přehledně zobrazuje oddělený sinusový signál a vygenerovaný šum. Ve druhém grafu je zobrazen histogram četností šumu, na kterém je vidět charakteristický zvonovitý tvar Gaussovy křivky. Třetí graf prezentuje konečný superponovaný signál vzniklý sečtením předchozích dvou složek.

Na závěr jsem provedl ověření správnosti. Pomocí výpočtu kvadratického průměru jednotlivých polí dat jsem zjistil skutečnou efektivní hodnotu vygenerovaného signálu i šumu a z nich určil dosažené hodnoty SNR. Výslednou hodnotu jsem vždy zobrazil v legendě posledního grafu.
